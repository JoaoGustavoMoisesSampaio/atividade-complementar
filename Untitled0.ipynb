Teste 1 – Fundamentos Matemáticos

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Carregar o arquivo CSV
dados = pd.read_csv("base_custos_transporte.csv")

# Mostrar a tabela
dados
```



##Etapa 2 – Identificar as variáveis

A base de dados possui duas variáveis:

 mes: número do mês;
custo_transporte`: custo total de transporte no mês, em reais.

### a) Variável independente

A variável independente é o mês, pois representa o tempo e é utilizada para analisar a variação dos custos.

### b) Variável dependente

A variável dependente é o custo de transporte, pois seu valor varia de acordo com o mês.

### c) Representação no contexto

A variável mes representa o período analisado, enquanto custo_transporte representa o valor gasto pela empresa com transporte em cada mês.



##Etapa 3 – Construir um gráfico

Será construído um gráfico de dispersão utilizando o mês no eixo x e o custo de transporte no eixo y.

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    dados["mes"],
    dados["custo_transporte"]
)

plt.title("Custos de Transporte por Mês")
plt.xlabel("Mês")
plt.ylabel("Custo de Transporte (R$)")
plt.grid(True)

plt.show()
```

### Análise do gráfico

Observando o gráfico, é possível perceber que os custos de transporte apresentam uma tendência de crescimento ao longo dos meses.

Os valores aumentam de forma geral do mês 1 até o mês 12, apesar de pequenas diferenças na velocidade desse crescimento entre alguns meses.

## Etapa 4 – Encontrar uma função

Será utilizada uma função afim no formato:

**f(x) = ax + b**

A função será encontrada utilizando os dados dos 12 meses e uma aproximação linear.

```python
# Encontrar os coeficientes da função afim
a, b = np.polyfit(
    dados["mes"],
    dados["custo_transporte"],
    1
)

print("Valor de a:", a)
print("Valor de b:", b)

print("\nFunção encontrada:")
print(f"f(x) = {a:.2f}x + {b:.2f}")
```

### Resultado da função

O coeficiente **a** representa aproximadamente quanto o custo de transporte varia a cada mês.

O coeficiente **b** representa o valor estimado da função quando x = 0.

A função encontrada representa aproximadamente a tendência dos custos de transporte observados nos 12 meses.

## Etapa 5 – Interpretar a função

### a) O que representa o coeficiente a?

O coeficiente **a** representa a variação média aproximada do custo de transporte a cada mês.

### b) O que representa o coeficiente b?

O coeficiente **b** representa o valor que a função estima para o custo quando o mês é igual a zero. Como não existe mês 0 dentro dos dados analisados, esse valor serve principalmente como parte matemática da função.

### c) O valor de a indica crescimento ou redução?

Como o valor de **a é positivo**, ele indica que os custos de transporte apresentam uma tendência de **crescimento** ao longo dos meses.

## Etapa 6 – Representar a função graficamente

Agora será adicionada ao gráfico dos dados reais a reta correspondente à função encontrada.

```python
# Criar valores para representar a reta
x = np.linspace(1, 15, 100)

# Calcular os valores da função
y = a * x + b

plt.figure(figsize=(10, 6))

# Dados reais
plt.scatter(
    dados["mes"],
    dados["custo_transporte"],
    label="Dados reais"
)

# Reta da função
plt.plot(
    x,
    y,
    label="Função afim"
)

plt.title("Custos de Transporte e Função Afim")
plt.xlabel("Mês")
plt.ylabel("Custo de Transporte (R$)")
plt.grid(True)
plt.legend()

plt.show()
```

## Etapa 7 – Fazer previsões

Utilizando a função afim encontrada, serão estimados os custos de transporte para os meses 13, 14 e 15.

```python
# Meses que serão previstos
meses_futuros = [13, 14, 15]

# Calcular as previsões
previsoes = []

for mes in meses_futuros:
    custo = a * mes + b
    previsoes.append(custo)

    print(f"Mês {mes}:")
    print(f"f({mes}) = {a:.2f} × {mes} + {b:.2f}")
    print(f"Custo estimado = R$ {custo:.2f}")
    print()
```

## Etapa 8 – Analisar o resultado

### a) A função encontrada representa bem a tendência dos dados?

Sim. A função representa aproximadamente a tendência geral dos dados, pois a reta acompanha o crescimento observado nos custos durante os 12 meses.

### b) Os custos estão aumentando ou diminuindo?

Os custos estão **aumentando** ao longo do período analisado.

### c) Aproximadamente quanto o custo aumenta a cada mês?

O aumento médio aproximado por mês é dado pelo coeficiente **a** da função encontrada.

Esse valor pode ser observado no resultado calculado pelo Python.

```python
print(f"O custo aumenta aproximadamente R$ {a:.2f} por mês.")
```

## Comparação entre os últimos valores observados e as previsões

O último valor observado na base corresponde ao mês 12. A partir da função afim, foram realizadas previsões para os meses 13, 14 e 15.

A tendência observada nos dados indica continuidade do crescimento dos custos, e a função matemática permite realizar uma estimativa dos valores futuros com base nesse comportamento histórico.
